```python
import pandas as pd

human_df = pd.read_csv(
    "../data/processed/amazon_labeling_100.csv"
)

llm_df = pd.read_csv(
    "../data/processed/amazon_llm_labels.csv"
)

print("Human-labeled:", len(human_df))
print("LLM-labeled:", len(llm_df))
```

```python
llm_training = llm_df.rename(
    columns={"llm_intent": "intent"}
).copy()

human_training = human_df.copy()

print("Human columns:")
print(human_training.columns.tolist())

print("\nLLM columns:")
print(llm_training.columns.tolist())
```

```python
training_df = pd.concat(
    [
        human_training[["tweet_id", "created_at", "text", "intent"]],
        llm_training[["tweet_id", "created_at", "text", "intent"]]
    ],
    ignore_index=True
)

print("Total training examples:", len(training_df))
```

```python
duplicates = training_df[
    training_df["tweet_id"].duplicated(keep=False)
]

print("Duplicate tweet IDs:", len(duplicates))

if len(duplicates) > 0:
    display(duplicates.sort_values("tweet_id"))
```

```python
print(
    training_df["intent"]
    .value_counts()
    .sort_values(ascending=False)
)
```

```python
training_path = "../data/processed/amazon_training_data.csv"

training_df.to_csv(
    training_path,
    index=False
)

print(f"Saved training dataset to: {training_path}")
print(f"Training examples: {len(training_df)}")
```
