```python
from collections import Counter
import pandas as pd
```

```python
full_path = r"C:\Users\sujat\.cache\kagglehub\datasets\thoughtvector\customer-support-on-twitter\versions\10\twcs\twcs.csv"
```

```python
brand_response_counts = Counter()

for chunk in pd.read_csv(
    full_path,
    usecols=["author_id", "inbound"],
    chunksize=100_000
):
    brand_rows = chunk[chunk["inbound"] == False]
    brand_response_counts.update(
        brand_rows["author_id"].astype(str)
    )

brand_response_counts.most_common(20)
```
